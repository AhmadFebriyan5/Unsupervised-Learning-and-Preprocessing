# Chapter 3: Unsupervised Learning and Preprocessing



## 1. Types of Unsupervised Learning



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris, make_blobs, make_moons
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.metrics import adjusted_rand_score
```

## 2. Preprocessing and Scaling


```python
iris = load_iris()
X = iris.data
y = iris.target

print("Feature names:", iris.feature_names)
print("Dataset shape:", X.shape)
print("First five samples:\n", X[:5])
```

## 3. Different Kinds of Preprocessing



```python
standard_scaler = StandardScaler()
X_standardized = standard_scaler.fit_transform(X)

minmax_scaler = MinMaxScaler()
X_minmax = minmax_scaler.fit_transform(X)

print("Standardized data (first five rows):")
print(X_standardized[:5])
print("\nMin-max scaled data (first five rows):")
print(X_minmax[:5])
```

## 4. Applying Data Transformations



```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].scatter(X[:, 0], X[:, 1], c=y)
axes[0].set_title("Original Data")
axes[0].set_xlabel("Sepal length")
axes[0].set_ylabel("Sepal width")
axes[1].scatter(X_standardized[:, 0], X_standardized[:, 1], c=y)
axes[1].set_title("Standardized Data")
axes[1].set_xlabel("Sepal length (scaled)")
axes[1].set_ylabel("Sepal width (scaled)")
plt.tight_layout()
plt.show()
```

## 5. Dimensionality Reduction with Principal Component Analysis (PCA)



```python
X_scaled = StandardScaler().fit_transform(X)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Explained variance ratio:")
print(pca.explained_variance_ratio_)
print("Total explained variance:", pca.explained_variance_ratio_.sum())
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y)
plt.xlabel("First principal component")
plt.ylabel("Second principal component")
plt.title("PCA on the Iris Dataset")
plt.show()
```

## 6. Clustering



### 6.1 k-Means Clustering



```python
X_blobs, y_blobs = make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=0)

kmeans = KMeans(n_clusters=3, n_init=10, random_state=0)
kmeans_labels = kmeans.fit_predict(X_blobs)

plt.figure(figsize=(7, 5))
plt.scatter(X_blobs[:, 0], X_blobs[:, 1], c=kmeans_labels)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], marker="^", s=150, edgecolor="black")
plt.title("k-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()

print("ARI:", adjusted_rand_score(y_blobs, kmeans_labels))
```

### 6.2 Agglomerative Clustering



```python
agglomerative = AgglomerativeClustering(n_clusters=3)
agglomerative_labels = agglomerative.fit_predict(X_blobs)

plt.figure(figsize=(7, 5))
plt.scatter(X_blobs[:, 0], X_blobs[:, 1], c=agglomerative_labels)
plt.title("Agglomerative Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()

print("ARI:", adjusted_rand_score(y_blobs, agglomerative_labels))
```

### 6.3 DBSCAN



```python
dbscan = DBSCAN(eps=0.8, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_blobs)

plt.figure(figsize=(7, 5))
plt.scatter(X_blobs[:, 0], X_blobs[:, 1], c=dbscan_labels)
plt.title("DBSCAN Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()

n_clusters = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
n_noise = np.sum(dbscan_labels == -1)
print("Number of clusters:", n_clusters)
print("Number of noise points:", n_noise)
print("ARI:", adjusted_rand_score(y_blobs, dbscan_labels))
```

## 7. Comparing and Evaluating Clustering Algorithms



```python
X_moons, y_moons = make_moons(n_samples=200, noise=0.05, random_state=0)
X_moons_scaled = StandardScaler().fit_transform(X_moons)

algorithms = [KMeans(n_clusters=2, n_init=10, random_state=0), AgglomerativeClustering(n_clusters=2), DBSCAN()]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, algorithm in zip(axes, algorithms):
    clusters = algorithm.fit_predict(X_moons_scaled)
    ax.scatter(X_moons_scaled[:, 0], X_moons_scaled[:, 1], c=clusters, s=30)
    ax.set_title(f"{algorithm.__class__.__name__}\nARI: {adjusted_rand_score(y_moons, clusters):.2f}")
    ax.set_xticks(())
    ax.set_yticks(())
plt.tight_layout()
plt.show()
```

## Kesimpulan

Preprocessing ini membantu menyiapkan data sebelum digunakan oleh algoritma machine learning. StandardScaler dan MinMaxScaler bisa digunakan buat menyamakan skala fitur. PCA dapat mengurangi jumlah dimensi sekaligus mempertahankan sebagian besar variasi data. Sementara itu, K-Means, Agglomerative Clustering, dan DBSCAN bisa digunakan buat menemukan kelompok data tanpa label target.
